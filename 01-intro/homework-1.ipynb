```python
first_median = df["horsepower"].median()
print(first_median)
```

```python
most_frequent_value = df["horsepower"].mode()[0]
print(most_frequent_value)
```

```python
df["horsepower"] = df["horsepower"].fillna(most_frequent_value)
```

```python
final_median = df["horsepower"].median()
print(final_median)
```

```python
df_asia = df[df["origin"] == "Asia"]
X = df_asia[["vehicle_weight", "model_year"]].head(7).to_numpy()
print(X)
```

```python
XTX = X.T@X
print(XTX)
```

```python
inverted_XTX = np.linalg.inv(XTX)
print(inverted_XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(y)
```

```python
w = inverted_XTX @ X.T @ y
print(w)
```

```python
sum_result = w.sum()
print(sum_result)
```
